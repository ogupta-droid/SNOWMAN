```python
import random
def intro():
    print("Lets play SNOWMAN")
    difficulty = input("What level difficulty do you want, easy, medium, or hard?")
    language = input("English, Spanish, or French?")
    return difficulty, language


EEwordbank = ['beach', 'ghost', 'truck', 'pizza', 'clock'] #Used Ai to get words
EMwordbank = ['elephant', 'mountain', 'football', 'dinosaur', 'sandwich'] #Used Ai to get words
EHwordbank = ['architecture', 'championship', 'grandparents', 'thanksgiving', 'passionfruit'] #Used Ai to get words
SEwordbank = ['playa', 'perro', 'libro', 'mundo', 'verde'] #Used Ai to get words
SMwordbank = ['aventura', 'guitarra','invierno', 'mariposa', 'teléfono'] #Used Ai to get words
SHwordbank = ['arquitectura', 'característica', 'computadora', 'universidad', 'extraordinario'] #Used Ai to get words
FEwordbank = ['avion', 'fleur', 'heure', 'jeune', 'route'] #Used Ai to get words
FMwordbank = ['chocolat', 'bicyclet', 'histoire', 'papillon', 'vacances'] #Used Ai to get words
FHwordbank = ['architecture', 'caracteristique', 'ordinateur', 'universite', 'extraordinaire'] #Used Ai to get words

def wordpick(lng, difficulty):
  if difficulty.lower() == "easy" and lng.lower() == "english":
    word = random.choice(EEwordbank)
  elif difficulty.lower() == "medium" and lng.lower() == "english":
    word = random.choice(EMwordbank)
  elif difficulty.lower() == "hard" and lng.lower() == "english":
    word = random.choice(EHwordbank)
  elif difficulty.lower() == "easy" and lng.lower() == "spanish":
    word = random.choice(SEwordbank)
  elif difficulty.lower() == "medium" and lng.lower() == "spanish":
    word = random.choice(SMwordbank)
  elif difficulty.lower() == "hard" and lng.lower() == "spanish":
    word = random.choice(SHwordbank)
  elif difficulty.lower() == "easy" and lng.lower() == "french":
    word = random.choice(FEwordbank)
  elif difficulty.lower() == "medium" and lng.lower() == "french":
    word = random.choice(FMwordbank)
  elif difficulty.lower() == "hard" and lng.lower() == "french":
    word = random.choice(FHwordbank)
  #Google told me how to use random

  guess = len(word)
  if lng.lower() == "french":
    guess += 6
  elif lng.lower() == "spanish":
    guess += 4
  elif lng.lower() == "english":
    guess += 3
  return word, guess


def showword(word, guessed_letters, hint_letters):
  display = ""
  for i in range(len(word)):
    if word[i] in guessed_letters or i in hint_letters:
      display += word[i] + " "
    else:
      display += "_ "
  return display
     # Google was used in function to explain how to do the display

while True:
  difficulty, language = intro()
  word, guess = wordpick(language, difficulty)
  #I was stuck on how to pull out my variables from the functions after a few runs so I googled it

  guessed_letters = []
  points = 0
  hint_used = False
  hint_letters = []

  print()
  print("Your word has", len(word), "letters.")
  print(f"You have {guess} guesses.")

  while guess > 0:

    display = showword(word, guessed_letters, hint_letters)

    print()
    print("Word:", display)
    print("Guesses left:", guess)
    print("Points:", points)

    if "_" not in display:  #Used google, how to figure out if there was a _ in it
       print()
       print("You Won!")
       print("The word was:", word)
       print("Points:", points)
       break

    print()
    print("Enter ONE letter to guess a letter.")
    print("Or enter the whole word if you think you know it.")
    print("Type 'hint' to get the first 2 letters.")
    print("A hint costs 2 guesses and 2 points.")
    print("The earlier you guess the whole word, the bigger your bonus!")
    print("WARNING: If your whole-word guess is wrong, you lose!")

    choice = input("Your guess: ").lower()

    if choice == "hint":
      if hint_used == False:
        if guess > 2:
          print("Hint: ", word[:2])

          hint_letters.append(0)
          hint_letters.append(1)

          guess -= 2
          points -= 2
          hint_used = True
        else:
          print("You dont have enough guesses for a hint.")
      else:
        print("You already used your hint.")
    elif len(choice) == 1 and choice.isalpha(): #Used google to figure out a way to not hardcode the entire alphabet agaion :(
      if choice in guessed_letters:
        print("You already guessed that letter.")
        points -= 5
        guess -= 1
      else:
        guessed_letters.append(choice)

        if choice in word:
          print("Correct!")
          points += 10
        else:
          print("Incorrect")
          points -= 5
        guess -= 1
    elif len(choice) > 1 and choice.isalpha():
      if choice == word:
        letters_left = 0
        for letter in word:
          if letter not in guessed_letters:
            letters_left += 1
        bonus = letters_left * 15
        points += bonus

        print()
        print("Correct! You guessed the whole word!")
        print(f"You had {letters_left} letters left.")
        print(f"Bonus: + {bonus} points!")
        print("You Won!")
        print(f"you got {points} points!")
        break
      else:
        print()
        print("Wrong you lose!")
        print(f"The word was:{word}")
        points -= 50

        guess = 0
        break
    else:
      print("Please enter a letter, a word, or 'hint'.")
  if guess == 0:
    print()
    print("Game over!")
    print(f"The word was: {word}")
    print(f"You got, {points}, points.")

  while True:
    try:
      play_again = int(input("Would you like to play again? yes (1) or no (2)?"))

      if play_again != 1 and play_again != 2:
        raise ValueError # looked up how to have it be repeated if it isn't an answer on Google

      break

    except ValueError:
      print("Please enter 1 or 2.")

  if play_again == 2:
    print('Thanks for playing SNOWMAN!')
    break
```
